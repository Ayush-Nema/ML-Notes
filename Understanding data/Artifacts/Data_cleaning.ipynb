# missing values in the `df`



```python
df = df.replace({"": pd.NA, "NA": pd.NA})
display(df.isna().sum())
```

# Remove the confirmed duplicates

```python
print("shape of df before duplicate removal:", len(df))

print("Extra identical rows:", df.duplicated().sum())


df = df.drop_duplicates().copy()
print("shape of df after duplicate removal:", len(df))
```

# Clean string and category labels

```python
_name = df.loc[0, "name"]
_name
```

```python
df["name"] = df["name"].astype("string").str.strip()
print(df["name"])
```

```python
df["phone"] = df["phone"].astype("string")
```

```python
df["city"] = df["city"].astype("string").str.strip().str.lower()
df["city"] = df["city"].replace("bangalore", "bengaluru")
```

```python
df["course"] = df["course"].astype("string").str.strip().str.lower()
df["course"] = df["course"].replace({"ml": "machine learning", "m.l": "machine learning"})
display(df[["student_id", "name", "city", "course", "phone"]])
```

# Convert numbers, then check their meaning

```python
age_before = df["age"].copy()
df["age"] = pd.to_numeric(df["age"], errors="coerce")
display(df["age"], age_before)
display(age_before[age_before.notna() & df["age"].isna()])
```

## Fix age

```python
bad_age = df["age"].notna() & ~df["age"].between(18, 100)  # getting bad age candidates

display(df["age"][bad_age])
display(df.loc[bad_age, ["student_id", "age"]])

df.loc[bad_age, "age"] = float("nan")
df["age"]
```

# Fee column

```python
fee_text = df["fee"].str.replace("₹", "", regex=False).str.replace(",", "", regex=False)
df["fee"] = pd.to_numeric(fee_text, errors="coerce")
display(fee_text[fee_text.notna() & df["fee"].isna()])

df.loc[df["fee"] < 0, "fee"] = float("nan")
display(df["fee"])
```

# Score column

```python
df["score"] = pd.to_numeric(df["score"], errors="coerce")
# bad_score = df["score"].notna() & ((df["score"] < 0) | (df["score"] > 100))
bad_score = df["score"].notna() & ~df["score"].between(0, 100)
display(bad_score)
display(df.loc[bad_score, ["student_id", "name", "score"]])

df.loc[bad_score, "score"] = float("nan")
display(df[["student_id", "age", "fee", "score"]])
```

# Clean booleans and datetimes

```python
active_text = df["is_active"].astype("string").str.strip().str.lower()
df["is_active"] = active_text.map({
    "yes": True, "y": True, "true": True, "1": True,
    "no": False, "n": False, "false": False, "0": False
}).astype("boolean")

display(df[["student_id", "is_active"]])
```

```python
date_text = df["enrolled_at"].astype("string").str.strip().str.replace("/", "-", regex=False)
df["enrolled_at"] = pd.to_datetime(date_text, format="%Y-%m-%d %H:%M", errors="coerce")

# NaT is pandas' missing datetime marker. February 30 is not a real date.
display(date_text[date_text.notna() & df["enrolled_at"].isna()])
display(df[["student_id", "enrolled_at"]])
```

# Data filling

### Age, city, course

```python
age_before = df["age"].copy()
age_median = df["age"].median()
print("Median of the valid ages:", age_median)
df["age"] = df["age"].fillna(age_median).astype("Int64")
age_after = df["age"]
l = pd.DataFrame({"Age Before": age_before, "Age After": age_after})
display(l)

df["city"] = df["city"].fillna("unknown").astype("category")
df["course"] = df["course"].astype("category")

display(df.isna().sum())
```

### fees

```python
df["fee_needs_review"] = (df["fee"] > 20000).astype("boolean")
df.loc[df["fee"].isna(), "fee_needs_review"] = pd.NA  # Cannot assess an unknown fee.
display(df.loc[df["fee_needs_review"].fillna(False), ["student_id", "fee"]])
```
